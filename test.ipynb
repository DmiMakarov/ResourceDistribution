```python
#в каком виде можно хранить эту информацию:
#1. Словарь - заказ, количество
#2. Структура - заказ, количество (минус в том, что могут стакаться несколько заказов)
#Можно сделать приоритетную очередь, в которой бы номер заказа
#можно сделать словарь имя заказа: количество деталей
#проссумируем количество деталей в словаре
#сравним с максимальным количеством деталей
#если всего деталей меньше, чем может быть сделано, то просто берём все детали
#если всего деталей больше, то можно посчитать по процентам. Вопрос - как
#Допустим, всего можно сделать 10 деталей
#3 заказа - 5, 7, 3 - всего 15
# 3 4 3 должен получиться ответ
#Надо брать равномерно по количеству временных заказов
# Надо брать 10/3 из каждого заказа - только как?
# изи - надо отсортировать по количеству, потом брать min(x, max_/n). Если min == x,
#то n-=1
```

```python
n = 10
a = {"a": 5, "b": 7, "c": 3}
b = dict(sorted(a.items(), key=lambda item: item[1]))
len_ = len(b)
for i, val in b.items():
    used = min(val, np.floor(n / len_) )
    b[i] -= used
    n -= used
    len_ -= 1

b
```
